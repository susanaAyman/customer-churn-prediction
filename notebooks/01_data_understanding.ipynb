## 6. Investigate the TotalCharges dtype issue

```
# TotalCharges looks numeric but is stored as text — find out why
non_numeric = df[pd.to_numeric(df['TotalCharges'], errors='coerce').isna()]
non_numeric[['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges']]
```

These rows have a blank/whitespace string instead of a number — usually customers
with `tenure == 0` (brand new, so no total charges yet). This is why `df.isnull().sum()`
below will report 0 missing values for TotalCharges even though these blanks exist —
they're empty strings, not `NaN`. We will fix the dtype and decide how to handle these
rows during cleaning (Day 3), not today.
